```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import pandas as pd
```

```python
client_id = "c90ea70f713742c69defa9541f372d06"
client_secret = "bfa4e194648b47d6a84da1ba3f9026b3"
auth_manager = SpotifyClientCredentials(client_id=client_id, client_secret=client_secret)
sp = spotipy.Spotify(auth_manager=auth_manager)
```

```python
spoti_dataset='https://open.spotify.com/playlist/2X4gLfwrZcZLKG8gU6QtdH?si=50a73baddab14b3a'
results = sp.playlist_tracks(spoti_dataset)
artist_ids=[]
for item in results['items']:
    artists = item['track']['artists']
    for artist in artists:
        artist_ids.append(artist['id'])

# Remove duplicates
artist_ids = list(set(artist_ids))
```

```python
for i in range(0, len(artist_ids), 50):
    batch = artist_ids[i:i+50]
    artists_info = sp.artists(batch)['artists']

    for artist in artists_info:
        artist_data.append({
            'name': artist['name'],
            'popularity': artist['popularity'],
            'followers': artist['followers']['total'],
            'genres': ", ".join(artist['genres']),
            'spotify_url': artist['external_urls']['spotify']
        })

# Create DataFrame
df = pd.DataFrame(artist_data)

# Sort by popularity
df = df.sort_values(by='popularity', ascending=False).reset_index(drop=True)
df = df.drop_duplicates(subset='name')
df
```
